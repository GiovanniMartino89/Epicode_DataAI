```python
#Esercizio 5. Slicing e metodi combinati
#Dati. vendite [120, 340, 275, 410, 90, 560, 205] e testo "Analisi vendite trimestre"

#1. su vendite eseguire separatamente vendite [::2] e vendite [::-1]. 
#Stampare entrambi i risultati

#Def. Slicing, è il recupero di un sottoinsieme di elementi con la sintassi
# lista[start:stop:step], restituisce sempre una nuova sequenza dello stesso tipo

vendite = [120, 340, 275, 410, 90, 560, 205]
print(vendite) 

print(vendite[::2]) #un elemento ogni 2
print(vendite[::-1]) #al contrario, lista capovolta

#2 Sulla stessa lista, eseguire separatamente vendite [:3] e vendite [:4]. 
#Stampare entrmabi i risultati e il numero di elementi di ciascuno

print(vendite[::3]) #ogni 3
print(vendite[::4]) # ogni 4

print(len(vendite[::3]))
print(len(vendite[::4]))

#3 su testo, applicare in sequenza strip(), poi lower() sul risultato ottenuto, poi split()
# sul risultato precedente. Stampare l'output dopo ciascuno dei tre passaggi, non solo
#il risultato finale

testo = "analisi vendite trimestre"

print(testo.strip(testo)) #rimuovi tutto
print(testo.strip()) #stampa esattamentente quello che c'è

print(testo.lower())

print(testo.split())

#Definizione richiesta. scrivere la definizione di slicing e spiegare, in base ai risultati 
#del punto 1, cosa cambia nella sequenza restituita quando il parametro step è 2 rispetto 
#a quando è -1

#criteri di completamento: sono stampati tutti i risultati richiesti dai 3 punti, 
#incluso l'output intermedio dopo ogni singolo metodo del punto 3
```
