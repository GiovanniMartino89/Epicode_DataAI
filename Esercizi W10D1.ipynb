```python
#Esercizio 1. Array NumPy
#Dati: temperature = [18, 22, 19, 25, 21] in gradi Celsius da rappresenaare e analizzare in NumPy

#1 Importare NumPy con l'alias np e creare un array a partire dalla lista temperatura

temperature = [18, 22, 19, 25, 21]

import numpy as np

np.array(temperature)

#2 Convertire l'array in gradi Farenheit applicando la formula (temperatura *9/5) + 32 in un'unica operazione vettoriale senza for

t = np.array(temperature)

conversione = (t * 9/5) + 32

print(conversione)

#3 calcolare media, minimo, massimo dell'array originale in Celsius, usando metodi statistici in NumPy

np.mean(temperature)
np.min(temperature)
np.max(temperature)

print(np.mean(temperature))
print(np.min(temperature))
print(np.max(temperature))

#Definizione richiesta: scrivere la definizione di Array NumPy e di operazione vettoriale 
#spiegando perché la conversione del punto 2  non richieda un ciclo esplicito

#Un Array NumPy è una struttura dati che memorizza elementi dello stesso tipo in una griglia multidimensionale
#non serve il ciclo perché l'array moltiplica ogni elemento singolarmente senza ciclo

#Criterio di completamento: stampare l'array convertito in Far e i tre valori del punto 3
```

```python
#Esercizio 2. Pandas Series
#Dati: un elenco di 5 città con la rispettiva popolazione, da rappresentare con una Series Panda

#1 importare Pandas con alias standard pd e creare una series con le popolazioni come valori e i nomi delle città come indice

import pandas as pd

popolazione = pd.Series(
    [2750000, 1350000, 910000, 850000, 630000],
    index=["Roma", "Milano", "Napoli", "Torino", "Palermo"]
)

print(popolazione)

#2 Accedere al valore della città utilizzando la sua etichetta indice non la posizione numerica

print(popolazione["Milano"])

print(popolazione["Palermo"])


#3 Creare una seconda Series a partire da un dizionario citta:popolazione e verificare che il risultato sia come al punto 1

popolazione = pd.Series({"Roma":2750000, "Milano":1350000, "Napoli":910000, "Torino":850000, "Palermo":630000}
)

print(popolazione)

#Definizione richiesta: def di Series e spiegare la differenza tra un indice della series e posizione elemento in una lista

#Una panda series è un array monodimensionale etichettato, contiene dati e indice associato a ciascun valore
#l'etichetta indice permette di assegnare etichette personalizzate e permette di accedere a dati tramite quello

#Criterio di completamento: stampare entrambe le Series create e il valore ottenuto per l'etichetta punto 2
```
