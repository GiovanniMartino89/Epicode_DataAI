```python
#Esercizio 2. Operatori e precedenza
#Valore_prezzo=45.90, quantita=3, sconto_percentuale=10, nome='mario', cognome='rossi'

#1. Calcolare il totale dell'ordine (prezzo * quantità) e poi applicare lo sconto, usando le
#parentesi per far calcolare prima il totale e poi la percentuale

#Def. Un operatore è un simbolo che esegue una operazione su uno o più valori, operandi, 
#e restituscie un risultato
#con le stringhe, ad esempio, + concatena * ripete (non moltiplica)

valore_prezzo = 45.90
quantita = 3
sconto_percentuale = 10
nome = "mario"
cognome = "rossi"

totale = (valore_prezzo * quantita)
print(totale)

totale_sconto = (totale) * (1 - (sconto_percentuale/100))
print(totale_sconto)

#uso di // e %. Il pritmo fa una definizione intera (arrotonda), % resto della divisione
# facciamo un esempio

totale_intero = valore_prezzo // quantita
print(totale_intero) #mi da 15

totale_modulo = valore_prezzo % quantita
print(totale_modulo)

#2. costruire una stringa di saluto concatenando nome e cognome con l'operatore +
#inserendo uno spazio tra i due valori

nome_cog = nome + cognome
print(nome_cog)
nome_cog = nome +' '+ cognome
print(nome_cog)
saluto= "ciaoooo" + "a" + "tutti"
print(saluto)
saluto_spazio = "ciao" + " " + "a" + " " + "tutti" + " " + "da" + " " + "mario rossi"
print(saluto_spazio)

#3.Usare un operatore di assegnazione composto (+=, *=, o -=) per aggiornare la variabile 
#quantita, simulando l'aggiunta di un prodotto all'ordine 

quantita = 3
quantita +=4
print(quantita)

sconto_percentuale = 10
sconto_percentuale -= 5
print(sconto_percentuale)

#Definzione richiesta. Scrivere la definizione di operatore e spiegare perché //e % restituisce
#risultati diversi da / sullo stesso paio di operandi

#Criterio di completamento. Il risultato è stampato con Print(), scelta parentesi motivata

```
