```python
#Esercizio 1. Array NumPy
#Dati: temperature = [18, 22, 19, 25, 21] in gradi Celsius da rappresenaare e analizzare in NumPy

#1 Importare NumPy con l'alias np e creare un array a partire dalla lista temperatura

temperature = [18, 22, 19, 25, 21]

import numpy as np

np.array(temperature)

#2 Convertire l'array in gradi Farenheit applicando la formula (temperatura *9/5) + 32 in un'unica operazione vettoriale senza for

t = np.array(temperature)

conversione = (t * 9/5) + 32

print(conversione)

#3 calcolare media, minimo, massimo dell'array originale in Celsius, usando metodi statistici in NumPy

np.mean(temperature)
np.min(temperature)
np.max(temperature)

print("Media", np.mean(temperature))
print(np.min(temperature))
print(np.max(temperature))

#Definizione richiesta: scrivere la definizione di Array NumPy e di operazione vettoriale 
#spiegando perché la conversione del punto 2  non richieda un ciclo esplicito

#Un Array NumPy è una struttura dati che memorizza elementi dello stesso tipo in una griglia multidimensionale
#non serve il ciclo perché l'array moltiplica ogni elemento singolarmente senza ciclo

#Criterio di completamento: stampare l'array convertito in For e i tre valori del punto 3
```

```python
#Esercizio 6. Masking (avanzato • 30 minuti)
#Dati: il DataFrame ordini, con colonne importo e citta.

#1. costruire una maschera booleana che individui gli ordini con importo superiore a una soglia a scelta, e usarla per filtrare il
#DataFrame con locl].

import pandas as pd

dati = {
    "id_ordine": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110, 111, 112],
    "cliente":   ["Rossi", "Bianchi", "Verdi", "Neri", "Gallo", "Conti",
                  "Rossi", "Bianchi", "Esposito", "Greco", "Verdi", "Conti"],
    "importo":   [120.50, 75.00, None, 230.00, 45.90, None,
                  310.00, 89.99, 150.00, None, 62.50, 199.00],
    "citta":     ["Torino", "Milano", "Roma", "Napoli", "Torino", "Milano",
                  "Roma", "Torino", "Napoli", "Milano", "Roma", "Torino"]
}
ordini = pd.DataFrame(dati)

maschera_bool = ordini["importo"] > 100

ordini_100 = ordini.loc[maschera_bool]

print(maschera_bool)
print(ordini_100)


#2 costruire un filtro che combini due condizioni con l'operatore &: 
#importo superiore alla soglia scelta al punto 1 e citta uguale a una città a scelta.

and_cond = ordini.loc[(ordini["importo"]>100) & (ordini["citta"]=="Torino")]

print(and_cond)

#3 costruire un filtro che usi l'operatore | per selezionare gli ordini di due città diverse, 
#poi negare lo stesso filtro con l'operatore ~ e confrontare il numero di righe dei due risultati.

cond = (ordini["citta"] == "Torino") | (ordini["citta"] == "Roma")

_loc = ordini.loc[cond]
_tilde = ordini.loc[~cond]

print("ordine due citta", len(_loc))          #7 righe
print("negazione", len(_tilde)) #righe

#Definizione richiesta: scrivere la definizione di masking 
#e spiegare perché su una Series o un DataFrame si usano gli operatori & e l, e non and/or.

#Il Masking è una tecnica che applica una condizione a un'intera colonnna o series: ottiene una sequenza di valori booleani
#true e false, riga per riga; gli operatori and/or in Python non funzionano elemento per elemento, come invece & e \

#Criterio di completamento: sono stampati i risultati dei tre filtri richiesti, 
#e la somma delle righe del filtro con | e del suo negato con ~ corrisponde al numero totale di righe del DataFrame.
```
