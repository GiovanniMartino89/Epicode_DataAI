```python
#Esercizio 2. Ciclo For

#Dati. articoli = ["penna", "quaderno", "zaino"], prezzi = {"penna": 1.5, "quaderno": 3.0, "zaino": 25.0}

#1. Usare un ciclo For per stampare ogni elemento della lista articoli, uno per riga

articoli = ["penna", "quaderno", "zaino"]

for articolo in articoli:
    print(articolo)

#2. Usare un ciclo for con un metodo . items() per stampare ogni coppia articolo-prezzo del dizionario prezzi

prezzi = {"penna": 1.5, "quaderno": 3.0, "zaino": 25.0}

for nome, prezzo in prezzi.items():
    print(nome, prezzo)

    
#3. All'interno del ciclo sul dizionario, usare un if per stampare un messaggio aggiuntivo solo per gli articoli prezzo > 10

prezzi = {"penna": 1.5, "quaderno": 3.0, "zaino": 25.0}

for nome, prezzo in prezzi.items(): 
    if prezzo > 10:
        print("prezzo alto", nome, prezzo)

#Def. richiesta scrivere la definizione di ciclo For e spiegare la differenza tra iterare direttamente su un dizionario e 
#iterare con .items()

#Def. Il ciclo For è un costrutto di iterazione che esegue lo stesso blocco una volta per ciascun elemento (passa in rassegna)
# dal primo all'ultimo elemento. Il metodo .items() è l'unico che restituisce la coppia chiave:valore, else solo chiavi

# Cr. Completamento: sono stampati tutti gli articoli, tutte le coppie articolo prezzo, messaggio aggiunto solo per art sopra soglia
```

```python
#Esercizio 5. List Comprehension
#Dati. Voti = [5,8,6,9,4,7,10]

#1 scrivere un ciclo for con .append() che costruisca una nuova lista contente solo i voti maggiori o uguali a 6

Voti = [5,8,6,9,4,7,10]
Voti_magg = []

for voto in Voti: 
    if voto >= 6:
        Voti_magg.append(voto)
print(Voti_magg)

#2 scrivere una list compr che ottenga lo stesso risultato del punto 1, con la sintassi espressione for elemento in iterabile
#if condizione

Voti_magg = [voto for voto in Voti if voto >= 6]
print(Voti_magg)

#3 scrivere una seconda list compr. che moltiplichi per 10 ciascun voto della lista voti, senza condizione filtro

Voti_molt = [voto * 10 for voto in Voti]
print(Voti_molt)

#. Def. richiesta scrivere la def di list compr ed elencare i casi in cui è preferibile tornare al ciclo for 

# Una List comprehension è una sintassi compatta che costruisce una nuova lista applicando un'espressione a ogni elemento
# di un iterabile, dove è possibile filtrare gli elementi con una condizione 
#1. Quando servono più istruzioni, gestione di errori o logica complessa

# Cr. compl. i risultati punto 1 e 2 sono stampati e coincidono; punto 3 stessa lunghezza voti
```

```python
#Esercizio 6. Funzioni e librerie 
#Dati un elenco di 5 numeri a scelta, da elaborare con funzioni  che usa una libreria della standard library

#1 importare la libreria math e definire una funzione radice_arrotondata(numero, decimale=2) che resistuisca con return
# la radice quadrata di numero, arrotondata al numero di decimali indicato con round()

import math 

def radice_arrotondata(numero, decimale =2):
    return round(math.sqrt(numero), decimale)

radice_arrotondata(68,2)

#2 richiamare la funzione su ciascuono dei 5 numeri usando un ciclo for, stampando ogni risultato

numeri = [68, 27, 32, 40, 60]

for numero in numeri:
    output = radice_arrotondata(numero)
    print(output)

#3 richiamare la funzione una volta senza indicare l'argomento decimali e una volta indicandolo esplicitamenente 
#come argomento keywords con un valore diverso da 2, confrontando i due output

radice_arrotondata(68) #8.25

radice_arrotondata(68, decimale = 4) #8.24


# Def. scrivere la definizione di libreria spiegando perché il comando import è necessario, anche per funzioni 
# della standard librari come math

# Cri. completamento: la funzione è richiamta correttamente sui 5 numeri nel ciclo for e il confronto tra chiamata default
# e chiamta keywords mostrano due output con numero diverso di decimali 
```
