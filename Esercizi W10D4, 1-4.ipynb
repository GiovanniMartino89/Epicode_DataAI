```python
#Esercizio 3 • Filtri con condizioni multiple
#Dati: il DataFrame prodotti dell'esercizio precedente, o uno equivalente con almeno 6 righe e una colonna categoria.

#1. costruire un filtro che selezioni i prodotti con prezzo superiore a una soglia a scelta 
#e categoria uguale a un valore a scelta, combinando le due condizioni con l'operatore &.

import pandas as pd #df Prodotti

Prodotti = pd.DataFrame({
    "nome" : ["caffettiera", "bollitore", "padella", "pentola", "bicchieri", "microonde"],
    "categoria" : ["pentolame", "elettrodomestici", "pentolame", "pentolame", "stoviglie", "elettrodomestici"],
    "prezzo": [20, 40, 35, 60, 15, 100]
})

Prodotti

condizione_prezzo = Prodotti["prezzo"] > 30
condizione_categoria = Prodotti["categoria"] == "pentolame"

filtro_prodotti = Prodotti.loc[condizione_prezzo & condizione_categoria]

print(filtro_prodotti)
```

```python
#2. usare .isin() per selezionare i prodotti che appartengono a una lista di due categorie a scelta, senza scrivere due condizioni
#unite da |. 

import pandas as pd

Prodotti[Prodotti["categoria"].isin(["pentolame", "stoviglie"])]
```

```python
#3. negare il filtro del punto 2 con l'operatore ~ per ottenere i prodotti che non appartengono a quelle categorie.

import pandas as pd

Prodotti[~Prodotti["categoria"].isin(["pentolame", "stoviglie"])]

Prodotti[]
```

```python
#Definizione richiesta: scrivere la definizione di .isin() e spiegare perché in Pandas si usano & e | 
#invece di and e or per combinare condizioni.

#.isin() è un metodo che verifica se ogni valore di una colonna è presente nella lista data
#In Pandas gli operatori logici & e \ sono equivalenti a and e or che si usano solo in Python puro

#Criterio di completamento: sono stampati i risultati dei tre filtri, 
#e la somma delle righe dei filtri dei punti 2 e 3 corrisponde al numero totale di righe del DataFrame.

condizione_prodotti = Prodotti["categoria"].isin(["pentolame", "stoviglie"]) #lo verifico tramite Len

filtro = Prodotti.loc[condizione_prodotti]
filtro_tilde = Prodotti.loc[~condizione_prodotti]

print(len(filtro) + len(filtro_tilde))   # 6
print(len(Prodotti))                      # 6
print(len(filtro) + len(filtro_tilde) == len(Prodotti))
```

```python
#Esercizio 4 • Strategie di imputazione
#Dati: una Series temperature con sei valori, di cui uno nullo in posizione intermedia, ad esempio (20, 21, None, 24, 25, 23].

#1. applicare separatamente fillna() con un valore fisso a scelta, ffill() e .interpolate() 
#alla stessa Series, ottenendo tre risultati distinti.

import pandas as pd

temperature = pd.Series([20, 21, None, 24, 25, 23])

temperature

print(temperature.ffill()) #valore precedente, 21
print(temperature.fillna(28))#valore fisso, 28
print(temperature.bfill())  #24 il valore successivo disponibile
print(temperature.interpolate()) #22.5 fa la media, interpolazione
```

```python
#3. indicare, per un caso in cui il valore nullo fosse il primo elemento della Series invece che uno intermedio, 
#quale dei quattro metodi non riuscirebbe a riempirlo, motivando la risposta.


import pandas as pd

temperature = pd.Series([None, 21, 22, 24, 25, 23])

temperature

print(temperature.ffill()) # ffill non riesce a riempirlo perché consindere il valore precedente per riempire il gap
# non essendoci un valore precedente, non riesce e dare un valore al primo elemento della series

print(temperature.interpolate()) #anche interpolate perché lui stima un punto intermedio  tra il precedente e il successivo
#essendo il primo, non può lavorare e dunque non riesce a riempire il gap, rimanendo None
```
