```python
#Esercizio 2. Ciclo For

#Dati. articoli = ["penna", "quaderno", "zaino"], prezzi = {"penna": 1.5, "quaderno": 3.0, "zaino": 25.0}

#1. Usare un ciclo For per stampare ogni elemento della lista articoli, uno per riga

articoli = ["penna", "quaderno", "zaino"]

for articolo in articoli:
    print(articolo)

#2. Usare un ciclo for con un metodo . items() per stampare ogni coppia articolo-prezzo del dizionario prezzi

prezzi = {"penna": 1.5, "quaderno": 3.0, "zaino": 25.0}

for nome, prezzo in prezzi.items():
    print(nome, prezzo)

    
#3. All'interno del ciclo sul dizionario, usare un if per stampare un messaggio aggiuntivo solo per gli articoli prezzo > 10

prezzi = {"penna": 1.5, "quaderno": 3.0, "zaino": 25.0}

for nome, prezzo in prezzi.items(): 
    if prezzo > 10:
        print("prezzo alto", nome, prezzo)

#Def. richiesta scrivere la definizione di ciclo For e spiegare la differenza tra iterare direttamente su un dizionario e 
#iterare con .items()

#Def. Il ciclo For è un costrutto di iterazione che esegue lo stesso blocco una volta per ciascun elemento (passa in rassegna)
# dal primo all'ultimo elemento. Il metodo .items() è l'unico che restituisce la coppia chiave:valore, else solo chiavi

# Cr. Completamento: sono stampati tutti gli articoli, tutte le coppie articolo prezzo, messaggio aggiunto solo per art sopra soglia
```

```python
#Esercizio 3. Funzioni con paramentri e return 
#Dati. una funziona da scrivere per calcolare il totale di un ordine, prezzo unitario e quantità

#1. definire una funziona calcola_totale (prezzo, quantità) che restituisca con return il prodotto delle due quantità

def calcola_totale(prezzo, quantità):
    return (prezzo * quantità)

#2. richiamare una volta la funzione con argomenti posizionali, e una volta con keywords, ordine invertito e stampare entrambi

print(calcola_totale(25.0,5)) #valori posizionali

#keywords
keywords = calcola_totale(prezzo = 25, quantità = 3)

#ordine invertito
invertito = calcola_totale (quantità = 3, prezzo = 25)

print(keywords)
print(invertito)

#3. modificare la funzione aggiungendo un parametro sconto con valore di deafult 0, che riduce il totale della percentuale 
#indicata quando viene passato un valore diverso da 0

def calcola_sconto(prezzo, quantita, sconto=0):
    totale = prezzo * quantita
    if sconto !=0:
        totale = totale - totale * sconto/100
        return totale
    
#Def. scrivere la def. di funziona con valore default, indicando quando l'argomento sconto può essere omesso nella chiamata

# Def. Una funzione è un blocco di condice organizzato e riutilizzabile che esegue un compito specifico
# con valore default vuol dire che è un paramentro nella funzione che verrà chiamato in automtico se la chiamata nalla funzione
#non specifica un valore diverso, può essere opzionale in quanto la funzione usa un valore di default 

# Cr. completamento: la funzione è richiamta almeno tre volte, posizionale, keywords, sconto, messaggio aggiuntivo solo sopra soglia
```
